```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

import time
import matplotlib.pyplot as plt
```

```python
# Transform: convert images to tensors and normalize to [0,1]
transform = transforms.Compose([
    transforms.ToTensor()
])

# Training dataset
train_dataset = datasets.MNIST(root='./data', train=True, transform=transform, download=True)

# Test dataset
test_dataset = datasets.MNIST(root='./data', train=False, transform=transform, download=True)

# Data loaders (batching)
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

print("Train samples:", len(train_dataset))
print("Test samples:", len(test_dataset))
```

```python
class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        # 1 input channel (grayscale), 32 filters, 3x3 kernel
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.relu = nn.ReLU()

        # after 2x pooling: 28x28 -> 14x14 -> 7x7
        self.fc1 = nn.Linear(64 * 7 * 7, 128)
        self.fc2 = nn.Linear(128, 10)  # 10 classes (0–9)

    def forward(self, x):
        x = self.relu(self.conv1(x))   # [batch, 32, 28, 28]
        x = self.pool(x)               # [batch, 32, 14, 14]
        x = self.relu(self.conv2(x))   # [batch, 64, 14, 14]
        x = self.pool(x)               # [batch, 64, 7, 7]
        x = x.view(x.size(0), -1)      # flatten
        x = self.relu(self.fc1(x))
        x = self.fc2(x)
        return x

model = CNN()
print(model)
```

```python
def train_model(model, device, train_loader, optimizer, criterion, epochs=1):
    model.to(device)
    model.train()

    start_time = time.time()

    for epoch in range(epochs):
        running_loss = 0.0
        correct = 0
        total = 0

        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item()

            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

        epoch_loss = running_loss / len(train_loader)
        epoch_acc = 100 * correct / total
        print(f"Epoch [{epoch+1}/{epochs}] | Loss: {epoch_loss:.4f} | Accuracy: {epoch_acc:.2f}%")

    end_time = time.time()
    print(f"Training time on {device}: {end_time - start_time:.2f} seconds\n")
    return end_time - start_time

def test_model(model, device, test_loader, criterion):
    model.to(device)
    model.eval()

    total_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)
            total_loss += loss.item()

            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    avg_loss = total_loss / len(test_loader)
    accuracy = 100 * correct / total
    print(f"Test Loss: {avg_loss:.4f} | Test Accuracy: {accuracy:.2f}%")
    return avg_loss, accuracy
```

```python
device_cpu = torch.device("cpu")
device_gpu = torch.device("cuda" if torch.cuda.is_available() else "cpu")

criterion = nn.CrossEntropyLoss()

# -------- CPU TRAINING --------
print("Training on CPU...")
model_cpu = CNN()
optimizer_cpu = optim.Adam(model_cpu.parameters(), lr=0.001)
cpu_time = train_model(model_cpu, device_cpu, train_loader, optimizer_cpu, criterion, epochs=1)
cpu_test_loss, cpu_test_acc = test_model(model_cpu, device_cpu, test_loader, criterion)

# -------- GPU TRAINING --------
print("\nTraining on GPU (CUDA)..." if torch.cuda.is_available() else "\nGPU not available, training on CPU again...")
model_gpu = CNN()
optimizer_gpu = optim.Adam(model_gpu.parameters(), lr=0.001)
gpu_time = train_model(model_gpu, device_gpu, train_loader, optimizer_gpu, criterion, epochs=1)
gpu_test_loss, gpu_test_acc = test_model(model_gpu, device_gpu, test_loader, criterion)

print("\n=== SUMMARY ===")
print(f"CPU training time: {cpu_time:.2f} s, Test Accuracy: {cpu_test_acc:.2f}%")
print(f"GPU training time: {gpu_time:.2f} s, Test Accuracy: {gpu_test_acc:.2f}%")
```
